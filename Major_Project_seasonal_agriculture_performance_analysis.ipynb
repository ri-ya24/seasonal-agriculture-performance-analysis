### 7.6.1 Outlier Investigation for Agricultural Variables

Since the numerical variables have different scales, agricultural and environmental variables are examined separately to identify potential outliers more clearly.

```python
# Selected agricultural and environmental variables

agricultural_variables = [
    'Farm_Area_Hectares',
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Sunlight_Hours_Day',
    'Soil_pH',
    'Soil_Moisture_pct',
    'Nitrogen_kg_ha',
    'Phosphorus_kg_ha',
    'Potassium_kg_ha',
    'Fertilizer_kg_ha',
    'Pesticide_Litre_ha',
    'Seed_Quality_Score',
    'Yield_Tonnes_Ha',
    'Water_Used_m3',
    'Water_Efficiency_t_per_1000m3',
    'Disease_Pest_Risk_pct'
]

plt.figure(figsize=(18, 10))
df[agricultural_variables].boxplot(rot=90)
plt.title('Box Plot of Agricultural and Environmental Variables')
plt.ylabel('Values')
plt.tight_layout()
plt.show()
```

### 7.6.2 Identifying Potential Outliers Using the IQR Method

The Interquartile Range (IQR) method is used to identify observations that fall below the lower bound or above the upper bound of the expected range.

```python
# Identifying potential outliers using the IQR method

outlier_summary = []

for column in agricultural_variables:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    outlier_summary.append({
        'Variable': column,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Potential Outliers': len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    by='Potential Outliers',
    ascending=False
)
```

### 7.6.3 Outlier Assessment

The IQR method identified potential outliers in several numerical variables. The highest number of potential outliers was observed in Water Efficiency, Yield, and Water Used.

However, an IQR-based outlier does not necessarily represent an incorrect observation. In agricultural data, extreme values may represent genuine differences in farm size, resource usage, productivity, or environmental conditions.

Therefore, the identified observations are retained at this stage rather than being removed automatically. Further analysis will consider the context and distribution of these variables before making any data-removal decision.

### 9.3 One-Way ANOVA: Yield Across Seasons

One-Way ANOVA (Analysis of Variance) is used to determine whether the average crop yield differs significantly across the three agricultural seasons: Kharif, Rabi, and Zaid.

The null hypothesis assumes that the mean yield is the same across all seasons, while the alternative hypothesis assumes that at least one season has a different mean yield.

```python
# 9.3 One-Way ANOVA: Yield Across Seasons

from scipy.stats import f_oneway

kharif_yield = df[df['Season'] == 'Kharif']['Yield_Tonnes_Ha']
rabi_yield = df[df['Season'] == 'Rabi']['Yield_Tonnes_Ha']
zaid_yield = df[df['Season'] == 'Zaid']['Yield_Tonnes_Ha']

anova_stat, p_value = f_oneway(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("F-statistic:", anova_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: The difference in mean yield across seasons is statistically significant.")
else:
    print("Result: The difference in mean yield across seasons is not statistically significant.")
```

### Interpretation

The One-Way ANOVA test produced an F-statistic of 1.54 and a p-value of 0.214.

Since the p-value is greater than the significance level of 0.05, the null hypothesis is not rejected.

Therefore, the differences in mean crop yield across Kharif, Rabi, and Zaid seasons are not statistically significant based on this test.

Although Kharif has a higher average yield than Rabi and Zaid, the observed differences are not large enough to be considered statistically significant in this dataset.

This indicates that season alone may not be a strong determinant of average crop yield, and other factors such as crop type, environmental conditions, irrigation, and resource usage may also influence yield.

### 9.4 One-Way ANOVA: Profit Across Seasons

One-Way ANOVA is used to determine whether the average agricultural profit differs significantly across Kharif, Rabi, and Zaid seasons.

The null hypothesis assumes that the mean profit is the same across all three seasons, while the alternative hypothesis assumes that at least one season has a different mean profit.

```python
# 9.4 One-Way ANOVA: Profit Across Seasons

from scipy.stats import f_oneway

kharif_profit = df[df['Season'] == 'Kharif']['Profit_INR']
rabi_profit = df[df['Season'] == 'Rabi']['Profit_INR']
zaid_profit = df[df['Season'] == 'Zaid']['Profit_INR']

anova_stat, p_value = f_oneway(
    kharif_profit,
    rabi_profit,
    zaid_profit
)

print("F-statistic:", anova_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: The difference in mean profit across seasons is statistically significant.")
else:
    print("Result: The difference in mean profit across seasons is not statistically significant.")
```

### Interpretation

The One-Way ANOVA test produced an F-statistic of 34.29 and a p-value of 1.71 × 10⁻¹⁵.

Since the p-value is far below the significance level of 0.05, the null hypothesis is rejected.

Therefore, the mean agricultural profit differs significantly across Kharif, Rabi, and Zaid seasons.

The seasonal descriptive statistics showed that Kharif had the highest average profit, followed by Rabi, while Zaid recorded a negative average profit. The ANOVA result confirms that these differences in average profitability across seasons are statistically significant.

This indicates that season has a significant association with agricultural profitability in the dataset, although the test does not by itself identify which specific pairs of seasons differ from each other.

### 9.5 Tukey HSD Post-Hoc Test: Seasonal Profit Comparison

Since the One-Way ANOVA showed a statistically significant difference in mean profit across seasons, the Tukey HSD (Honestly Significant Difference) post-hoc test is performed to identify which specific pairs of seasons differ significantly.

The test compares Kharif vs Rabi, Kharif vs Zaid, and Rabi vs Zaid while controlling for multiple comparisons.

```python
# 9.5 Tukey HSD Post-Hoc Test: Seasonal Profit Comparison

from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey_result = pairwise_tukeyhsd(
    endog=df['Profit_INR'],
    groups=df['Season'],
    alpha=0.05
)

print(tukey_result)
```

### Interpretation

The Tukey HSD post-hoc test shows that all three pairs of seasons have statistically significant differences in mean profit.

The mean profit for Rabi is approximately ₹91,225 lower than Kharif, while the mean profit for Zaid is approximately ₹203,719 lower than Kharif.

Similarly, the mean profit for Zaid is approximately ₹112,494 lower than Rabi.

For all three comparisons, the `reject` value is `True`, indicating that the differences are statistically significant at the 5% significance level.

Therefore, agricultural profitability differs significantly between every pair of seasons in the dataset. Kharif shows the strongest profitability, followed by Rabi, while Zaid has the weakest profitability.

Overall, the Tukey HSD results provide further statistical evidence that seasonal conditions are associated with substantial differences in agricultural profitability.

# 15. Limitations

Although the analysis provides useful insights into seasonal agricultural performance, certain limitations should be considered while interpreting the results.

### 15.1 Dataset Coverage

The analysis is based on the available dataset containing observations from selected states, districts, crops, and seasons. Therefore, the findings may not represent agricultural conditions across all regions of India.

### 15.2 Dataset Distribution

The number of observations is not equally distributed across all seasons, crops, and geographical categories. This imbalance may influence comparisons between groups.

### 15.3 Missing Values

A small number of missing values were present in rainfall, soil moisture, and yield variables. These values were handled using median imputation, which may introduce some degree of estimation rather than using the original observations.

### 15.4 Correlation Does Not Imply Causation

Correlation analysis identifies relationships between variables but does not establish cause-and-effect relationships. For example, a positive relationship between water usage and yield does not necessarily mean that increasing water usage will directly increase yield.

### 15.5 Outliers

Several variables contained observations identified as potential outliers using the IQR method. These observations were retained because they may represent genuine agricultural conditions rather than data-entry errors. However, extreme values can influence averages and statistical results.

### 15.6 External Factors

The dataset does not capture every factor that can affect agricultural performance, such as detailed soil characteristics, farmer experience, machinery availability, government support, labour availability, transportation costs, or changing market conditions.

### 15.7 Statistical Limitations

The ANOVA analysis identifies whether group means differ statistically but does not explain all possible factors responsible for those differences. Other statistical or machine-learning methods could provide deeper insights.

### 15.8 Generalization of Findings

The results and recommendations should be interpreted within the context of the available dataset. Further validation using larger, real-world, and more diverse agricultural datasets would improve the reliability and generalizability of the findings.
